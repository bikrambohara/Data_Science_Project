```python
import pandas as pd 
import os
```

```python
print("Hello World")
```

```python
KEEP_COLUMNS = [
     "HomeTeam", "AwayTeam",
    "FTHG", "FTAG", "FTR",
    "HS", "AS", "HST", "AST"
]

RENAME_COLUMNS = {
    "Date": "date",
    "HomeTeam":"home_team",
    "AwayTeam":"away_team",
    "FTHG":"home_goals",
    "FTAG":"away_goals",
    "FTR": "result",
    "HS": "home_shots",
    "AS": "away_shots",
    "HST": "home_shots_target",
    "AST": "away_shots_target"
}

print(KEEP_COLUMNS)
```

```python
# Function to extract season from filename

def extract_season_from_filename(filename):
    season_part = filename.replace("season-", "").replace(".csv", "")

    start_year = "20" + season_part[:2]
    end_year= season_part[2:]

    return f"{start_year}_{end_year}"

```

```python
def load_and_clean_data(folder_path, league_name):
    all_seasons = []

    for file in os.listdir(folder_path):
        if file.endswith('.csv'):
            file_path = os.path.join(folder_path, file)
            df = pd.read_csv(file_path)

            # Keep only relevant columns
            df = df[KEEP_COLUMNS]
            df = df.rename(columns=RENAME_COLUMNS)

            df["date"] = pd.to_datetime(df["date"],dayfirst=True,errors='coerce' )
            df["season"] = extract_season_from_filename(file)
            df['League'] = league_name


            all_seasons.append(df)

    return pd.concat(all_seasons, ignore_index=True)
```

```python
def main():
    epl = load_and_clean_data('../data/raw/epl', "EPL")
    laliga = load_and_clean_data('../data/raw/laliga', "LaLiga")

    combined = pd.concat([epl, laliga], ignore_index=True)

    combined.to_csv('../data/processed_data/cleaned_football_data.csv', index=False)
    print("All seasons cleaned and combined succesfully")

if __name__ == "__main__":
    main()
```
