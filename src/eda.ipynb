```python
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns 
```

```python
# df = pd.read_csv("../data/processed_data/cleaned_football_data.csv")
df = pd.read_csv("cleaned_football_data.csv")
```

1. This plot shows that home wins occur more frequently than away wins, indicating a home advantage, which justifies including home-related features in the model.


```python
plt.figure(figsize=(10,6))
sns.countplot(x="result", data= df)
plt.title("Distribution of Match Results (H/D/A)")
plt.xlabel("Match Result")
plt.ylabel("Number of Matches")
plt.show()
```

2. This plot shows that Home teams tend to score slightly more goals on average, reinforcing the importance of home advantage

```python
plt.figure(figsize=(10,6))
sns.kdeplot(df["home_goals"], label="Home Goals", fill=True)
sns.kdeplot(df["away_goals"], label="Away Goals", fill=True)
plt.title("Goal Distribution: Home vs Away")
plt.xlabel("Goals Scored")
plt.legend()
plt.show()


print("Average Home Goals:", df["home_goals"].mean())
print("Average Away Goals:", df['away_goals'].mean())
```

3. This plot shows that different leagues show different win–draw patterns, which supports adding league information as a feature.

```python
# League wise 

plt.figure(figsize=(8,4))
sns.countplot(x="result", hue="League", data=df)
plt.title("Match Result Distribution by League")
plt.xlabel("Result")
plt.ylabel("Matches")
plt.show()
```

4. This plot shows that most football matches have around 2 to 3 goals, which is consistent with real football trends

```python
# Total goals per match 

df["total_goals"] = df["home_goals"] + df["away_goals"]

plt.figure(figsize=(10,6))
sns.histplot(df["total_goals"], bins=10, kde=True)
plt.title("Total Goals per Match Distribution")
plt.xlabel("Total Goals")
plt.ylabel("Matches")
plt.show()
```

5. This plot shows that Matches with higher shots on target generally result in wins, which validates using shots-on-target-based historical features

```python
# Shots on target vs results

plt.figure(figsize=(10,6))
sns.boxplot(x='result', y='home_shots_target', data=df)
plt.title("Home Shots on Target vs Match Result")
plt.xlabel("Match Result")
plt.ylabel("Home Shots on Target")
plt.show()
```
