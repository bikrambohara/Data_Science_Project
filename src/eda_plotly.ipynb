```python
import pandas as pd 
import plotly.express as px 
```

```python
df = pd.read_csv("../data/processed_data/cleaned_football_data.csv")
```

```python

fig = px.bar(
    df["result"].value_counts().reset_index(),
    x="result",
    y="count",
    labels={"result": "Match Result", "count": "Number of Matches"},
    title="Distribution of Match Results (H/D/A)"
)

fig.show()
```

```python
fig2 = px.histogram(
    df,
    x=["home_goals", "away_goals"],
    barmode="overlay",
    opacity=0.6,
    nbins=15,
    title="Home vs Away Goals Distribution",
    labels={"value": "Goals", "variable": "Team Type"}
)

fig2.update_layout(template="plotly_white")
fig2.show()
```

```python
league_result = df.groupby(["League", "result"]).size().reset_index(name="count")

fig3 = px.bar(
    league_result,
    x="result",
    y="count",
    color="League",
    barmode="group",
    title="Match Results by League",
    labels={"result": "Match Result", "count": "Number of Matches", "League": "League"}
)

fig3.update_layout(template="plotly_white")
fig3.show()
```

```python
df["total_goals"] = df["home_goals"] + df["away_goals"]

fig4 = px.histogram(
    df,
    x="total_goals",
    nbins=15,
    title="Total Goals per Match Distribution",
    labels={"total_goals": "Total Goals"}
)

fig4.update_layout(template="plotly_white")
fig4.show()
```

```python
fig5 = px.box(
    df,
    x="result",
    y="home_shots_target",
    color="result",
    title="Home Shots on Target vs Match Result",
    labels={"result": "Match Result", "home_shots_on_target": "Home Shots on Target"}
)

fig5.update_layout(template="plotly_white")
fig5.show()
```

```python
season_trend = df.groupby("season")["total_goals"].mean().reset_index()

fig6 = px.line(
    season_trend,
    x="season",
    y="total_goals",
    markers=True,
    title="Average Total Goals per Match by Season",
    labels={"total_goals": "Average Goals"}
)

fig6.update_layout(template="plotly_white")
fig6.show()
```

```python
season_league_trend = df.groupby(["season", "League"])["total_goals"].mean().reset_index()

fig6b = px.line(
    season_league_trend,
    x="season",
    y="total_goals",
    color="League",
    markers=True,
    title="Season-wise Goal Trend by League"
)

fig6b.update_layout(template="plotly_white")
fig6b.show()
```
