```python

import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../data/processed_data/cleaned_football_data.csv")
```

```python
df.head()
df.info()
```

```python
df = df.drop(columns=[
    'date'
])
```

conver result Object to Numbers

```python
df['result'] = df['result'].map({
    'Win': 2,
    'Draw': 1,
    'Loss': 0
})
```

Encode CATEGORICAL FEATURES(Columns to Encode)

```python
categorical_cols = ['home_team', 'away_team', 'season', 'League']
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

for col in categorical_cols:
    df[col] = le.fit_transform(df[col])
```

```python
df.dtypes
```

```python
df.columns
```

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score
```

```python
x= df.drop('result', axis=1)
y = df['result']

x_train, x_test, y_train, y_test = train_test_split(x,y, test_size=0.2, random_state=42)
```

```python
model = RandomForestClassifier(n_estimators=100)
model.fit(x_train, y_train)

pre=model.predict(x_test)
print("Accuract:", accuracy_score(y_test, y_pred=pre))
```
